```python
#part 2: sorting OLTP events in a data lake 
#reminder that a DATA LAKE "describes the process of 'tossing' large amonts of unstructured (OLTP) data into a big container"
import boto3

#creating S3 client 
s3 = boto3.client("s3")
```

```python
#name of new S3 bucket using my netID
bucket_name = "dsan6000-kmd347"

#just making the new S3 bucket
s3.create_bucket(Bucket=bucket_name) 

print("Created bucket:", bucket_name)
```

```python
#part 2.2: taking .parquet files stored locally in data folder and add them to subfolder
#importing path to work with local data folder 
from pathlib import Path 

#set path to local data folder 
data_dir = Path("data")

#loop through every paraquet file in data folder 
for file_path in data_dir.glob("*.parquet"): 

    #creating S3 path inside wikipedia-hourly subfolder using formatted string
    s3_key = f"wikipedia-hourly/{file_path.name}"

    #uploading parquet file to S3 bucket 
    s3.upload_file(
        str(file_path),
        bucket_name, 
        s3_key
    )

print("Uploaded all parquet files to S3.")
```
