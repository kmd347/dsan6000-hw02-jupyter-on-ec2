```python
import boto3
from pathlib import Path
```

```python
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)
```

```python
s3 = boto3.client("s3")
```

```python
#getting list of objects stored in hourly_parquet folder (in dsan6000-wikipedia S3 bucket)
response = s3.list_objects_v2(
    Bucket="dsan6000-wikipedia", 
    Prefix="hourly_parquet/"
)

#making empty list to store parquet file paths 
parquet_files = []

#looping through objects that were taken from S3
for obj in response["Contents"]:

    #seeing if object is a parquet file 
    if obj["Key"].endswith(".parquet"):

        #add parquet file path to list 
        parquet_files.append(obj["Key"])

print(len(parquet_files))

print(parquet_files[:5])
```

```python
#looping through each parquet file found in S3 bucket 
for key in parquet_files: 

    #just getting the file name from full S3 path
    filename = Path(key).name

    #making local path where file will be saved 
    local_path = data_dir / filename 

    #downloading parquet file from S3 into local data folder 
    s3.download_file(
        "dsan6000-wikipedia", 
        key, 
        str(local_path)
    )

#after all files finish downloading, print message 
print("Downloaded", len(parquet_files), "parquet files.")
```
